## Here I will demonstrated how to implement tools in LangGraph. 

After that -

- I will add this feature to the existing chatbot






![Tool node execution workflow](../images/9_7_tool_calling.png)



- Here we have two workflow of chatbot. One is the simple and another one has the tool calling feature. 

- The "chat_node" in the 2nd workflow has the dicision making capability. Depending on the user's query, it decides whether it requires a tool to call. 

- For decison making, I have added two alternative paths LLM to follow. One is normal forward path, another is ToolNode.
- This ToolNode contains tools/tool and then through which LLM calls the tools
- Tool condition : it resides inside the tool node, it is a pre-built conditional edge function that helps the graph to decide should it go to the tool node or back to the LLM or somewhere else.
